```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
pd.set_option("display.max_columns", 50)
sns.set_style("whitegrid")

# Load dataset
df = sns.load_dataset('titanic')
```

```python
# Numerical feature nnalysis analysis
num_cols = ["age", "fare", "sibsp", "parch"]
cat_cols = ["survived", "pclass", "sex", "embarked", "class", "who", "adult_male", "deck", "embark_town", "alone"]
```

```python
df[num_cols].head()
```

```python
# summary statistics
df[num_cols].describe()
```

```python
print("Mean:")
print(df[num_cols].mean())
print("-"*30)
print("Median:")
print(df[num_cols].median())
print("-"*30)
print("Mode:")
print(df[num_cols].mode().iloc[0])
```

```python
# Variance and standar deviation
print("Variance:")
print(df[num_cols].var())
print("-"*30)
print("standar deviation:")
print(df[num_cols].std())
```

```python
# Skewness value & Kurtosis
print("Skewness value:")
print(df[num_cols].skew())
print("-"*30)
print("Kurtosis:")
print(df[num_cols].kurtosis())
```

```python
for col in num_cols:
  plt.figure(figsize = (5,3))
  sns.histplot(df[col], kde=True)
  plt.title(f"Histogram Distiribution: {col}")
  plt.xlabel(f"{col}")
  plt.ylabel("frequency")
  plt.show()
```

```python
# Boxplots
for col in num_cols:
  plt.figure(figsize = (5,3))
  sns.boxplot(x = df[col])
  plt.title(f"Histogram Distiribution: {col}")
  plt.xlabel(f"{col}")
  plt.ylabel("frequency")
  plt.show()
```

```python
# Categorial Feature Analysis

df[cat_cols].head()
```

```python
# Frequency table
for col in cat_cols:
  print(df[col].value_counts())
  print(df[col].value_counts(normalize=True))
  print("-" * 50)
```

```python
# count plot
for col in cat_cols:
  plt.figure(figsize = (5,3))
  sns.countplot(x = df[col])
  plt.title(f"Count plot: {col}")
  plt.xlabel(f"{col}")
  plt.ylabel("Count")
  plt.show()
```
