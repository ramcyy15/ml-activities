```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
np.set_printoptions(suppress=True)
```

```python
# Coin Toss Simulation
n = 500
outcomes = np.random.choice(["H","T"], size = n)
print("Head:",outcomes[:5])
print("Length:",len(outcomes))

heads = np.sum(outcomes == "H")
tails = np.sum(outcomes == "T")

print("Heads:", heads)
print("Tails:", tails)
print("Estimated P(Heads):", heads/n)
print("Estimated P(Tails):", tails/n)

```

```python
# Dice role simulation
n = 1000
dice = np.random.randint(1,7, size = n)
print("Head:",dice[:5])
print("Length:",len(dice))
for p_dice in np.unique(dice):
  print(f"Probability of {p_dice}: {np.mean(dice == p_dice)}")

vals, counts = np.unique(dice, return_counts=True)
print(vals)
print(counts)
```

```python
plt.bar(vals,counts)
plt.title("Counts of Dice Outcomes")
```

```python
# Conditional probability
data = pd.DataFrame({
    "likes_ML": np.random.choice([1,0], 100, p=[0.6,0.4])
})

# DL depends on whether the student likes ML
data["likes_DL"] = [
    np.random.choice([1,0], p = [0.7, 0.3]) if ml else
    np.random.choice([1,0], p=[0.2,0.8])
                     for ml in data['likes_ML']
]

p_dl_given_ml = data[data["likes_ML"] == 1]['likes_DL'].mean()

print("P(ML):", round(data['likes_ML'].mean(),2))
print("P(DL):", round(data['likes_DL'].mean(),2))
print("P (DL | ML) :",round(p_dl_given_ml,2))
```

```python
# Distribution Shape
n = 5000
plt.figure(figsize = (4,3))
plt.hist(np.random.uniform(0,1,n), alpha = 1.0, bins = 20, edgecolor ="k")
plt.title("Uniform [0,1]")

plt.figure(figsize = (4,3))
plt.hist(np.random.normal(0,1,n), alpha = 1.0, bins = 20, edgecolor ="k")
plt.title("Normal N[0,1]")
```
