```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.set_printoptions(suppress=True)
```

```python
# Simple Line: y = 2x + 1

def line_func(x):
  return 2 * x +1
```

```python
x = np.linspace(-3, 3, 100)
y = line_func(x)
```

```python
print(x[:5])
print(y[:5])
```

```python
# give x & y values, find the teh slope of the line

plt.figure(figsize = (6,4))
plt.plot(x,y)
plt.title("Line Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(alpha = 0.3)
plt.show()

# choose two points on the line
x1, x2 = -2, 1
y1, y2 = line_func(x1), line_func(x2)
print(y1, y2)
```

```python
# curve y = x^2
def curve(x):
  return x ** 2
x_val = np.linspace(-3,3,100)
y_val = curve(x)
```

```python
# plot

plt.figure(figsize = (6,4))
plt.plot(x_val, y_val)
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(alpha = 0.3)
plt.show()
```

```python
# Estimating slope with small change in X | Slope = dy/dx

def slope_estimate(func, x1, h=0.0001):
  x2 = x1 + h
  y1 = curve(x1)
  y2 =  func(x2)
  dy = y2 - y1
  dx = h
  slope = dy /dx
  return slope
```

```python
points = [-2,0,2]
for p in points:
  s = slope_estimate(curve, p)
  print(f"x = {p} estimated slope = {s:.2f}")
```

```python
# Sample dataset
x = np.array([1,2,3])
y_true = np.array([2,4,6])

# f(x) = w * x
# y = w * x

# starting guess of w
w = 4.0
# learning rate
lr = 0.001

def loss(w):
  y_pred = w * x
  return np.mean((y_pred - y_true) ** 2)


# derivative of loss
def dloss(w):
  y_pred = w * x
  return np.mean(2 * x * (y_pred - y_true))


print("Start w: ", w, "Loss:",loss(w))

for step in range(30):
  slope = dloss(w)
  w = w - lr * slope
  print("Step", step + 1, "w:", round(w, 3), "loss:", round(loss(w),3))
```
