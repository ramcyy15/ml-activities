```python
v1 = np.array([1,2,3,4])
v2 = np.array([5,6,7,8])
print("v1:", v1)
print("v2:", v2)
vector_add = v1 + v2
print("Add:",vector_add)
vector_multi = v1 * v2
print("Multi:",vector_multi)

A = np.array([
    [1,2,3],
    [4,5,6]
])

B = np.array([
    [10,11,12],
    [13,14,15]
])

C = np.array([
    [1,2,3],
    [3,4,5],
    [6,7,9]
])

print(f"Shape A: {A.shape} B: {B.shape}")
A_add_B = A + B
print("A + B:", A_add_B)
A_T = A.T
print("A Transpose: ", A_T)
C_X = np.linalg.inv(C)
print("A inverse:", C_X)
```

```python
v1 = np.array([1,2])
v2 = np.array([3,4])
dot = np.dot(v1,v2)
print("Dot:", dot)
```

```python
# Matrix Multiplication and Linear Model

X = np.array([
    [1.0, 30.0],
    [2.0, 50.0],
    [3.0, 70.0]
])

print("X:\n", X)
print("X shape:", X.shape)
```

```python
w = np.array([
    [0.5],
    [0.1]
])

print("Weight Shape:", w.shape)
```

```python
# formula: y = wX + b
Xw = X @ w
print("Xw:\n", Xw)
```

```python
b = 2.0
y = Xw + b
print("y = Xw + b:\n", y)
```
