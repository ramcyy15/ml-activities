### Tasks

**Task 1: Data Loading and Inspection**
- Load the dataset into a pandas DataFrame.
- Display the first few rows and use `.info()` to check for missing values or data type issues.

**Task 2: Inventory Status**
- Create a new column named `in_stock` that is `True` if `quantity` > 0 and `False` otherwise.

**Task 3: Filtering**
- Filter the DataFrame to show only the books in the 'Sci-Fi' genre.
- Filter the DataFrame to find books that are currently out of stock.

**Task 4: Aggregation**
- Calculate the total value of the inventory (sum of `price * quantity`).
- Find the average price of books grouped by `genre`.

```python
import pandas as pd
```

```python
# Task 1: Data Loading and Inspection
# Load 'library_books.csv' into a DataFrame and display the first 5 rows and .info()
df = pd.read_csv("library_books.csv")
print(df)
```

```python
# Task 2: Inventory Status
# Create a new column 'in_stock' that is True if quantity > 0, otherwise False
df["in_stock"] = df["quantity"] > 0
print(df)
```

```python
# Task 3: Filtering
# 1. Filter to show only 'Sci-Fi' genre books
# 2. Filter to show only books where 'in_stock' is False
scifi_books = df[(df['genre'] == 'Sci-Fi') & (df['in_stock'])]
print(scifi_books)


```

```python
# Task 4: Aggregation
# 1. Calculate total inventory value (price * quantity)
# 2. Calculate average price grouped by 'genre'
total_inv = df["price"] * df["quantity"]

print(total_inv)

avg_price = df.groupby("genre")['price'].mean()
print(avg_price)
```

## Exercise 1 (Easy): Books Above Average Price

Identify all books whose `price` is greater than the average price of all books in the DataFrame.

```python
greater_avg = df["price"] > df["price"].mean()
print(df[greater_avg])
```

## Exercise 2 (Medium): Authors with Mixed Inventory Status

Identify authors who have books that are both available (`in_stock` is True) and unavailable (`in_stock` is False) in the DataFrame.

```python
mixed_inventory_authors = df.groupby('author')['in_stock'].nunique() > 1
mixed_authors = mixed_inventory_authors[mixed_inventory_authors].index.tolist()
print(mixed_authors)
```

## Exercise 3 (Hard): Top Author by Quantity per Genre (Post-1940)

For each `genre`, determine the `author` who has the highest total `quantity` of books, considering only books published after the year 1940.

```python
books_post_1940 = df[df['year_published'] > 1940]
genre_author_total_quantity = books_post_1940.groupby(['genre', 'author'])['quantity'].sum().reset_index()
top_authors_per_genre = genre_author_total_quantity.loc[genre_author_total_quantity.groupby('genre')['quantity'].idxmax()]
print(top_authors_per_genre)
```
