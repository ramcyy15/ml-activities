````python
import numpy as np
````

````python
vector_u_list = [1, 2, 3]
vector_v_list = [4, 5, 6]

matrix_A_list = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

matrix_B_list = [
    [9, 8, 7],
    [6, 5, 4],
    [3, 2, 1]
]
````

````python
# Task 1

oned_arr_u = np.array(vector_u_list)
oned_arr_v = np.array(vector_v_list)

print(oned_arr_u)
print(oned_arr_v)

print(np.array(matrix_A_list))
print(np.array(matrix_B_list))
````

````python
# Task 2

print(oned_arr_u + oned_arr_v)
print(np.dot(oned_arr_u, oned_arr_v))
print(np.linalg.norm(oned_arr_u))
````

````python
# Task 3

print(np.array(matrix_A_list) + np.array(matrix_B_list))
print(np.dot(np.array(matrix_A_list), np.array(matrix_B_list)))
print(np.array(matrix_A_list).transpose())
print(np.linalg.det(np.array(matrix_A_list)))
````

````python
# Task 4
print(np.linalg.inv(np.array(matrix_A_list)))
print(np.linalg.solve(np.array(matrix_A_list), np.array([10, 20, 30])))
````
