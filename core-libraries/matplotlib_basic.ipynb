```python
import matplotlib.pyplot as plt
```

```python
x = [0, 1, 2, 3, 4, 5]
y = [0, 2, 4, 6, 8, 10]
plt.figure(edgecolor="black")
plt.plot(x,y)
plt.title("Simple Linear Plot")
plt.xlabel("Time")
plt.ylabel("Value")
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y1 = [1, 4, 9, 16, 25]
y2 = [1, 8, 27, 64, 125]
plt.figure()
plt.plot(x, y1, color = "blue", linestyle = "dashed", label = "Square")
plt.plot(x, y2, color = "red", linestyle = "solid", marker = "s", label = "Cube")
plt.legend()
plt.title("Growth Comparison")
plt.grid(alpha = 0.3)
plt.show()
```

```python
categories = ['A', 'B', 'C']
values = [10, 20, 15]
plt.figure(figsize=(10, 4))

plt.subplot(1,2,1)
plt.bar(categories,values)
plt.title("Bar Chart")

plt.subplot(1,2,2)
plt.scatter([1,2,3,4,5],[5,4,3,2,1],color = "green")
plt.title("Scatter Plot")

plt.tight_layout()
plt.show()
```
