```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = sns.load_dataset('tips')
data
```

### Exercise 1: Easy (Distribution)
Create a histogram or a distribution plot for the `total_bill` column to visualize its spread. Experiment with the number of bins or adding a KDE (Kernel Density Estimate) line.

```python
sns.displot(data=data, x = 'total_bill', bins = 20, kde = True)
```

### Exercise 2: Medium (Relationships)
Create a scatter plot showing the relationship between `total_bill` (x-axis) and `tip` (y-axis). Use the `time` column to color the points (hue) and use different markers for `smoker` status.

```python
sns.scatterplot(data=data, x = "total_bill", y = "tip", hue = 'time', style = "smoker")
```

### Exercise 3: Hard (Categorical & Faceting)
Create a faceted grid (using `FacetGrid` or `relplot`/`catplot`) that shows a boxplot of `total_bill` across different `days`, separated into different columns based on the `time` of day (Lunch vs. Dinner), and further distinguished by `sex`.

```python
sns.catplot(
    data=data,
    x="day",
    y="total_bill",
    hue="sex",
    col="time",
    kind="box",
)
```
